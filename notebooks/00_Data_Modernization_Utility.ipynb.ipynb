```python
# ==============================================================================
# 00 | IN-PLACE TEMPORAL OVERWRITE PROTOCOL (+13 YEARS)
# 00 | PROTOCOLO DE SOBREESCRITURA TEMPORAL DIRECTA (+13 AÑOS)
# ==============================================================================
import os
import shutil
import pandas as pd

print("⏳ [SYSTEM] Initiating in-place temporal migration protocol...")

# 1. Definición de rutas base
RAW_DIR = '../data/raw'
PROCESSED_DIR = '../data/processed'

train_path = os.path.join(RAW_DIR, 'train.csv')
features_path = os.path.join(RAW_DIR, 'features.csv')
processed_path = os.path.join(PROCESSED_DIR, 'retail_data_processed.csv')

# Función auxiliar para respaldar y desplazar fechas
def shift_dataset_dates(file_path, date_col='Date', years_to_add=13):
    if not os.path.exists(file_path):
        print(f"⚠️ [WARNING] File not found: {file_path}. Skipping.")
        return
    
    # Crear respaldo .bak de seguridad si aún no existe
    backup_path = file_path + '.bak'
    if not os.path.exists(backup_path):
        shutil.copyfile(file_path, backup_path)
        print(f"📦 [BACKUP] Created safety copy: {backup_path}")

    # Leer dataset
    df = pd.read_csv(file_path)
    
    if date_col in df.columns:
        df[date_col] = pd.to_datetime(df[date_col]) + pd.DateOffset(years=years_to_add)
        # Si contiene Year, Month, Week calculados, actualizarlos también
        if 'Year' in df.columns:
            df['Year'] = df[date_col].dt.year
        if 'Month' in df.columns:
            df['Month'] = df[date_col].dt.month
        if 'Week' in df.columns:
            df['Week'] = df[date_col].dt.isocalendar().week.astype(int)
    elif 'Year' in df.columns:
        df['Year'] = df['Year'] + years_to_add
    
    # Sobreescribir archivo original con el mismo nombre
    df.to_csv(file_path, index=False)
    
    # Reporte
    years_detected = sorted(df['Year'].unique()) if 'Year' in df.columns else sorted(pd.to_datetime(df[date_col]).dt.year.unique())
    print(f"✅ [UPDATED] {os.path.basename(file_path)} ➔ Active Years: {years_detected}")

# 2. Ejecución sobre los archivos clave
print("\n--- Migrating Raw Sources ---")
shift_dataset_dates(train_path)
shift_dataset_dates(features_path)

print("\n--- Migrating Curated Processed Store ---")
shift_dataset_dates(processed_path)

print("\n" + "="*65)
print(" 🚀 MIGRATION SUMMARY / RESUMEN DE MIGRACIÓN")
print("="*65)
print(" 📅 Historical horizon 2010-2012 successfully shifted to 2023-2025.")
print(" 📁 Standard file names preserved with zero 'modernized' suffixes.")
print(" 🛡️ Backups (.bak) available in target folders for safety.")
print("="*65 + "\n")
```

```python
# ==============================================================================
# CALIBRACIÓN EXACTA AL HORIZONTE 2023-2025
# ==============================================================================
import os
import shutil
import pandas as pd

RAW_DIR = '../data/raw'
PROCESSED_DIR = '../data/processed'

train_path = os.path.join(RAW_DIR, 'train.csv')
features_path = os.path.join(RAW_DIR, 'features.csv')
processed_path = os.path.join(PROCESSED_DIR, 'retail_data_processed.csv')

def fix_to_target_horizon(file_path, target_start_year=2023):
    if not os.path.exists(file_path):
        return
    
    # 1. Si existe respaldo .bak original (2010), restaurar desde ahí
    bak_path = file_path + '.bak'
    if os.path.exists(bak_path):
        shutil.copyfile(bak_path, file_path)
        print(f"🔄 Restaurado desde copia base: {os.path.basename(bak_path)}")
    
    df = pd.read_csv(file_path)
    
    # Identificar columna temporal
    if 'Date' in df.columns:
        df['Date'] = pd.to_datetime(df['Date'])
        current_min_year = df['Date'].dt.year.min()
        offset_needed = target_start_year - current_min_year
        
        if offset_needed != 0:
            df['Date'] = df['Date'] + pd.DateOffset(years=offset_needed)
            print(f"📅 {os.path.basename(file_path)}: Desplazamiento de {offset_needed} años aplicado.")
            
        if 'Year' in df.columns:
            df['Year'] = df['Date'].dt.year
        if 'Month' in df.columns:
            df['Month'] = df['Date'].dt.month
        if 'Week' in df.columns:
            df['Week'] = df['Date'].dt.isocalendar().week.astype(int)
            
    elif 'Year' in df.columns:
        current_min_year = df['Year'].min()
        offset_needed = target_start_year - current_min_year
        df['Year'] = df['Year'] + offset_needed
        
    df.to_csv(file_path, index=False)
    years = sorted(df['Year'].unique()) if 'Year' in df.columns else sorted(df['Date'].dt.year.unique())
    print(f"✅ {os.path.basename(file_path)} bloqueado exactamente en: {years}\n")

# Ejecutar sobre los archivos clave
fix_to_target_horizon(train_path)
fix_to_target_horizon(features_path)
fix_to_target_horizon(processed_path)
```
